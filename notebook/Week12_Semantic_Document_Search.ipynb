```python
import chromadb
import pandas as pd
import os

print("Semantic Document Search Project")
print("All libraries imported successfully!")
```

```python
DOCUMENTS_PATH = "../dataset/documents"
DB_PATH = "../chroma_db"

client = chromadb.PersistentClient(path=DB_PATH)

embedding_function = chromadb.utils.embedding_functions.DefaultEmbeddingFunction()

collection = client.get_or_create_collection(
    name="documents",
    embedding_function=embedding_function
)

print("ChromaDB initialized successfully!")
print("Total documents:", collection.count())
```

```python
query = "How computers learn from data"

results = collection.query(
    query_texts=[query],
    n_results=3
)

print("Search Query:", query)
print("\nRelevant Documents:\n")

for doc, metadata, distance in zip(
    results["documents"][0],
    results["metadatas"][0],
    results["distances"][0]
):
    print("Document:", metadata["filename"])
    print("Content:", doc)
    print("Distance:", round(distance, 4))
    print("-" * 40)
```

```python
similarity_data = []

for doc, metadata, distance in zip(
    results["documents"][0],
    results["metadatas"][0],
    results["distances"][0]
):
    similarity_score = 1 / (1 + distance)

    similarity_data.append({
        "Document": metadata["filename"],
        "Distance": distance,
        "Similarity Score": similarity_score
    })

similarity_df = pd.DataFrame(similarity_data)

display(similarity_df)
```

```python
similarity_df.to_csv(
    "../dataset/notebook_similarity_scores.csv",
    index=False
)

print("Similarity scores saved successfully!")
```
